```python
# ask user for a number and display negative or positive
num = int(input('Enter a number: '))
if num < 0:
  print('Negative')
elif num == 0:
  print('You have entered 0')
else:
  print('Positive')
```

```python
# ask user for marks and print the grade
# 90+ → A
# 80-89 → B
# 70-79 → C
# 60-69 → D
# Below 60 → F
marks = int(input('Enter marks: '))
if marks >=90:
  print('A grade')
elif marks >= 80:
  print('B grade')
elif marks >= 70:
  print('C grade')
elif marks >= 60:
  print('D grade')
else:
  print('Fail')
```

```python
# ask user for temperature and diplay the weather
# 40 or above → Very Hot
# 30-39       → Hot
# 20-29       → Pleasant
# Below 20    → Cold
temp = int(input('Enter temperature: '))
if temp >= 40:
  print('Very Hot')
elif temp >= 30:
  print('Hot')
elif temp>= 20:
  print('Pleasant')
else:
  print('Cold')
```
