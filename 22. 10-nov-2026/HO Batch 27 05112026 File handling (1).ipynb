```python
file = open('subjects.txt','r')
subjects = file.readlines()  # create a list
print(subjects)
file.close()
```

```python
# remove \n from the name of each subject
new_subject = []
for subject in subjects:
  subject = subject.strip()
  new_subject.append(subject)
new_subject
```

```python
new = []
for s in subjects:
  new.append(s.replace('\n',''))
new
```

```python
# file append
file = open('notes.txt','a')
file.write('\nI can now save this information')
file.close()
```

```python
# three modes of file handling
# 'r': read existing information
# 'w': write, replacing existing content
# 'a': appends information at the end
```
