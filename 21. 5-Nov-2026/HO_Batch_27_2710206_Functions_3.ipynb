```python
def analyze_marks(marks):
  total = 0
  highest = marks[0]
  lowest = marks[0]
  for mark in marks:
    total += mark
    if mark > highest:
      highest = mark
    if mark < lowest:
      lowest = mark
  return total, highest, lowest
```

```python
marks = [75,88,62,91,70]
total,highest,lowest = analyze_marks(marks)
print(total)
print(highest)
print(lowest)
```

```python
# create a fn analyze_numbers
# return total, number of even numbers, odd numbers
# numbers = [10, 15, 22, 31, 40, 51]
def analyze_number(numbers):
  total = even = odd = 0
  for number in numbers:
    total += number
    if number % 2 == 0:
      even += 1
    else:
      odd += 1
  return total,even,odd
```

```python
numbers = [10, 15, 22, 31, 40, 51]
tot, ev, od = analyze_number(numbers)
print(tot,ev,od)
```

```python
def get_high_marks(marks):
  high_marks = []
  for mark in marks:
    if mark>=80:
      high_marks.append(mark)
  return high_marks
```
