```python
# ATM Withdrawal
# ASk user for account balance and withdrawal amount
# if withdrawal amount > balance -> insufficient balance\
# otherwise transaction successful
# if remaining balance is below 1000:
#   warning: low balance
#   otherwise: transaction successful
balance = int(input('ENter account balance: '))
amount = int(input('ENter withdrawal amount: '))
if amount > balance:
  print('Insufficient balance')
elif balance - amount < 1000:
  print('You have withdrawn Rs.',amount)
  print('Low balance')
else:
  print('You have withdrawn Rs.',amount)
```

```python
# A cinema has three rules:
# customer must be 18 or older
# if they are 18+, they must have a valid ticket
#if they have a ticket, they can enter
age = int(input('Enter age: '))
ticket = input('DO you have a ticket: ')
if age >= 18 and ticket == 'yes':
  print('Entry allowed')
else:
  print('Entry denied')
```

```python
# A bank wants to check whether someone can get a loan.
# Rules:
# Age must be 21 or above.
# If age is eligible, ask monthly income.
# If income is at least Rs. 100,000, ask whether the
# person has existing loans.
# If they have no existing loans → eligible.
# Otherwise → manual review.
# If income is below Rs. 100,000 → not eligible.
# If age is below 21 → not eligible.
age = int(input('Enter age: '))
if age >= 21:
  income = int(input('What is your monthly income: '))
  if income >= 100000:
    status = input('Do you have previous loans: ')
    if status == 'yes':
      print('Manual review')
    else:
      print('Loan approved')
  else:
    print('Not eligible')
else:
  print('Not eligible')
```
