```python
python = {'Ali','Sara','Ahmed','Zoya'}
ai = {'Sara','Zoya','Hamza','Bilal'}
# union: combine both groups(sets)
all_students = python | ai
all_students
# | (pipe) sign is used for union in sets
all_students = python.union(ai)
all_students
```

```python
# Intersection: find shared members (common)
both_courses = python & ai
# & (ampersand) sign is used for intersection in sets
both_courses
both_courses = python.intersection(ai)
both_courses
```

```python
# Difference - find members in one group but not the other
python_only = python - ai
python_only
```

```python
# symmetric difference: find members who are in exactly one group
one_course_only = python ^ ai
one_course_only
```

```python
a = {1,2,3,4}
b = {3,4,5,6}
print(a | b) # union (every item without duplication)
print(a & b) # intersection (only common items)
print(a - b) # difference (items in a but not in b)
print(a ^ b) # symm diff: (only not common items)
```
