```python
print('==========================================\n'
      '           FEEDBACK ANALYZER              \n'
      '==========================================\n')

# Take feedback, remove extra spaces at the beginning/end,
# and convert the text to lowercase.
feedback = input('Enter your feedback: ').strip().lower()

# Calculate basic statistics.
char = len(feedback)
spaces = feedback.count(' ')

# Split feedback into a list of words.
word_list = feedback.split()
words = len(word_list)

# Initialize counters and longest-word variables.
vowels = 0
letters = 0
good = 0
bad = 0
longest_word_length = 0
longest_word = ''

# Count vowels and letters using a for loop.
for i in feedback:
    if i in 'aeiou':
        vowels += 1

    if i in 'abcdefghijklmnopqrstuvwxyz':
        letters += 1

# Analyze individual words.
for word in word_list:

    # Remove common punctuation from the start and end.
    clean_word = word.strip(".,!?;:\"'()[]{}#*/")

    # Count complete words, not substrings.
    if clean_word == 'good':
        good += 1

    if clean_word == 'bad':
        bad += 1

    # Find the longest word.
    if len(clean_word) > longest_word_length:
        longest_word_length = len(clean_word)
        longest_word = clean_word

# Display feedback statistics.
print('\n', feedback)
print('Number of characters:', char)
print('Number of words:', words)
print('Number of spaces:', spaces)
print('Number of letters:', letters)
print('Number of vowels:', vowels)
print('Longest word length:', longest_word_length)
print('Longest word:', longest_word)
print('"good" appears:', good)
print('"bad" appears:', bad)

# Determine the overall sentiment.
if good > 0 and bad == 0:
    print('\nOverall sentiment: Positive')
elif bad > 0 and good == 0:
    print('\nOverall sentiment: Negative')
elif good > 0 and bad > 0:
    print('\nOverall sentiment: Mixed')
else:
    print('\nOverall sentiment: Neutral')

print('==========================================')
```
