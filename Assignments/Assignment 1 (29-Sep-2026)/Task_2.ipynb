```python
days=["Monday", "Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
usage=[]
total=0
for i in range(len(days)):
  user = float(input("Enter your usage: "))
  usage.append(user)
  total += user

print("===================================",
      '\n       DATA USAGE REPORT',
      "\n===================================\n")

for i in range(len(days)):
  print(days[i],':',usage[i])
  average= total / len(days)

highest_usage = 0
heavy_usage_days = 0
for i in range(len(usage)):
  if usage[i] > highest_usage:
    highest_usage = usage[i]
    if usage[i] > 4:
      heavy_usage_days += 1 

print("\nTotal Usage", total)
print("Average Usage:", average)
print('Highest Daily Usage:', highest_usage)
print('Heavy Usage Days:', heavy_usage_days)
if total >= 30:
  print('Category: Heavy')
elif total >= 15 and total <30:
  print('Category: Moderate')
else:
  print("Category: Light")
```
