```python
# # ask user for a password
# if password contains only letters print that
# if passeword contains only numbers print that
# if password contains both, print password accepted
passw = input('Enter a password: ')
if passw.isalpha():
  print('Only letters')
elif passw.isdigit():
  print('Only numbers')
elif passw.isalnum():
  print('Password accepted')
else:
  print('Invalid password')
```

```python
# # ask user for employee ID
# Must start with "EMP"
# Remaining part must contain digits
# Total length must be 8 characters
# sample employee ID: EMP12345
emp = input('Enter employee ID: ')
if emp.startswith('EMP') and len(emp) == 8 and emp[3:].isdigit():
  print('Valid')
else:
  print('Invalid')
```
