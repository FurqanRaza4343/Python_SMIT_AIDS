```python
# using loops add numbers from 1 to 5
total = 0 #accumulator
for i in range(1,6):
  total = total + i
print(total)
# i = 1, total = 0, total = 0 + 1 -> 1
# i = 2, total = 1, total = 1 + 2 -> 3
# i = 3, total = 3, total = 3 + 3 -> 6
# i = 4, total = 6, total = 6 + 4 -> 10
# i = 5, total = 10, total = 10 + 5 -> 15
```

```python
# find the sum of all even numbers from 1 to 100
total = 0
for i in range(1,101):
  if i % 2 == 0:
    total = total + i
print(total)
```

```python
#another way to solve above problem
total = 0
for i in range(2,101,2):
  total = total + i
print(total)
```

```python
# count even numbers from 1 to 20
count = 0 #counter
for i in range(1,21):
  if i % 2 == 0:
    count = count + 1
print(count)
```

```python
# count numbers divisible by both 3 and 5 between 1 and 100
count = 0 #counter
for i in range(1,101):
  if i % 3 == 0 and i % 5 == 0:
    count = count + 1     # count += 1
print(count)
```

```python
# ask user for number of students, add the marks of each student
students = int(input('ENter number of students: '))
total =0
for i in range(students):
  marks = float(input('Enter marks: '))
  total = total + marks     # total += marks
print('TOtal:',total)
print('Avg:', total/students)
```

```python
# now add pass/fail count to the above problem, pass >=50
students = int(input('ENter number of students: '))
total = 0
passed = 0
failed = 0
for i in range(students):
  marks = float(input('Enter marks: '))
  total = total + marks     # total += marks
  if marks >= 50:
    passed += 1
  else:
    failed += 1
print('TOtal:',total)
print('Avg:', total/students)
print('Passed:',passed)
print('Failed:',failed)
```
