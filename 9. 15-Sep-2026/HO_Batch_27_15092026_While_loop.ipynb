```python
# ask user to enter a password, they have unlimited
# attempts, the program stops as soon as right password
# is entered
for i in range(5):
  password = input('Enter password: ')
  if password == 'python':
    print('Correct password')
```

```python
# print numbers from 10 to 1 using while loop
i = 10
while i >= 1:
  print(i)
  i = i - 1
```

```python
# print even numbers from 2 to 20 using while loop
i = 2
while i <= 20:
  print(i)
  i = i + 2
```

```python
# to the above code, inform user when he entres wrong password
password = ''
while password != 'python':
  password = input('Enter password: ')
  if password != 'python':
    print('Wrong password!!!')
print('Login successful')
```

```python
# keep asking user for a number until they enter a -ve num
number = 0
while number >= 0:
  number = int(input('Enter a number: '))
print('Negative number entered')
```

```python
# sum number 1 to 100 using while loop
total = 0
i = 1
while i <= 100:
  total = total + i
  i = i + 1
print(total)
```
