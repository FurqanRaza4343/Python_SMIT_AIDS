```python
numbers = [10, 20, 30, 40, 50, 60, 70, 80]

print(numbers[1:-1]) # [20,30,40,50,60,70]
print(numbers[-2:1:-1]) # [70,60,50,40,30]
print(numbers[::3]) # [10,40,70]
print(numbers[-1::-2]) # [80,60,40,20]
print(numbers[6:1:-2]) # [70,50,30]
# [10,20,30,40,50,60,70,80]
# [0,1,2,3,4.5.6.7]
```
