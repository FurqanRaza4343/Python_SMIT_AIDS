```python
# find the smallest number
smallest = int(input('ENter a number: '))
for i in range(4):
  number = int(input('ENter a number: '))
  if number < smallest:
    smallest = number
print('Smallest:', smallest)
```

```python
# ask user for a word and count number of 'a'
word = input('Enter a word: ')
count = 0
for letter in word:
  if letter == 'a':
    count += 1
print('Number of a:',count)
```

```python
#vowels: a,e,i,o,u
# ask user for a word and count vowels
word = input('Enter a word: ').lower()
count = 0
for letter in word:
  if letter == 'a' or letter == 'e' or letter == 'i' or letter == 'o' or letter == 'u':
    count += 1
print('Number of vowels:',count)
```

```python
# ask user for a word and display no. of characters,
# no. of vowels, no. of consonants, no. of a
word = input('Enter a word: ')
count = vowel = consonant = a = 0
for letter in word:
  count += 1
  if letter in 'aeiouAEIOU':
    vowel += 1
  else:
    consonant += 1
  if letter == 'a':
    a += 1
print('Number of characters:',count)
print('Number of vowels:',vowel)
print('Number of consonants:',consonant)
print('Number of a:', a)
```

```python
# ask user to input a password
# password must contain a number and @ sign
# notify user if above condition is not met
password = input('Enter a password: ')
number = False
sign = False
for i in password:
  if i in '0123456789':
    number = True
  if i == '@':
    sign = True
if number == True:
  if sign == True:
    print('Password accepted')
  else:
    print('Use @ sign in password')
else:
  print('Use a number in password')
```
