```python
marks = [75,88,62,91,70]
def analyze_marks(marks):
  total=0
  highest=marks[0]
  lowest=marks[0]
  for mark in marks:
    total += mark

    if mark > highest:
      highest = mark

    if mark < lowest:
      lowest = mark

  print('Total:',total)
  print('Highest:',highest)
  print('Lowest:',lowest)
```

```python
analyze_marks(marks)
```
