```python
# ask user for student name, maths, physics, chmistry
# and english marks. Each subject total marks are 100
# calculate total marks, avg marks, percentage
# display name, total, avg, percentage(with % sign)
#total = chemi + maths + eng + phy
# avg = total /4
# percentage = total / 400 * 100
name = input('Enter name: ')
math = float(input('Enter maths marks: '))
chem = float(input('Enter chem marks: '))
eng = float(input('Enter eng marks: '))
phy = float(input('Enter phy marks: '))
total = eng + phy + chem + math
avg = total / 4
percentage = total / 400 * 100
print('Name:',name)
print('Total:',total)
print('Average:',avg)
print('Percentage: ' + str(percentage) + '%')
```
