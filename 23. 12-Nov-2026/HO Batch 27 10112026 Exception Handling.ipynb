```python
# EXCEPTION HANDLING:
# exception handling lets us decide what our program
# should do when a particular runtime exception occurs
```

```python
number = int(input('Enter a whole nnumber: '))
print('You entered:',number)
```

```python
try:
  number = int(input('Enter a whole nnumber: '))
  print('You entered:',number)
except ValueError:
  print('That is not a valid whole number')
```

```python
try:
  first = float(input('Enter first number: '))
  second = float(input('Enter second number: '))
  result = first / second
except ValueError:
  print('Please enter valid number')
except ZeroDivisionError:
  print('The second number can not be 0')
else:
  print("Result:",result)
```

```python
try: # CHECKS THE CODE FOR EXCEPTION
  number = int(input('Enter a number: '))
except ValueError: # RUN IN CASE OF EXCEPTION
  print('Invalid number')
else: # RUNS IN CASE OF NO EXCEPTION
  print('Number',number)
finally: # ALWAYS RUNS WHETHER TTHERE IS EXCEPTION OR NOT
  print('Input attempt completed')
```

```python
try:
  number = int(input('Enter a number: '))
  result = 100 / number
  print(result)
except (ValueError, ZeroDivisionError) as e:
  print('Operation failed')
  print('Technical details',e)
```

```python
marks = int(input('Enter marks out of 100: '))
if marks < 0 or marks > 100:
  raise ValueError('Marks must be between 0 and 100')
print('Accepted marks:',marks)
# raise deliberately signals that a condtion is unacceptable
```

```python
try:
  marks = int(input('Enter marks out of 100: '))
  if marks < 0 or marks > 100:
    raise ValueError('Marks must be between 0 and 100')
  print('Accepted marks:',marks)
except ValueError as error:
  print('Invalid marks', error)
```
